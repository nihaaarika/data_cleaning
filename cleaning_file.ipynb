```python
# Convert all columns to numeric where possible
df = df.apply(pd.to_numeric, errors="coerce").fillna(0)

# Helper function: agar column exist nahi karta toh 0
def safe_col(df, colname):
    return df[colname] if colname in df.columns else 0

# 🔹 Fire / Property Insurance
df["Fire_Property_Insurance"] = safe_col(df, "Insurance") + safe_col(df, "Housing.Status") + safe_col(df, "Utility Bills")

# 🔹 Health Insurance (40% of Insurance)
df["Health_Insurance"] = safe_col(df, "Insurance") * 0.4  

# 🔹 Life Insurance (60% of Insurance)
df["Life_Insurance"] = safe_col(df, "Insurance") * 0.6  

# 🔹 Transport Related
df["Transport_Expenses"] = safe_col(df, "Transport") + safe_col(df, "Travel Expenses")

# 🔹 Accommodation
df["Accommodation"] = safe_col(df, "Rent") + safe_col(df, "Hotel")

# 🔹 Tax Related
df["Tax_Expenses_Final"] = safe_col(df, "Tax Expenses")

# 🔹 Travel Total
df["Travel_Total"] = safe_col(df, "Travel Expenses")

# 🔹 Family Related
df["Family_Total"] = safe_col(df, "Personal_Expenses") + safe_col(df, "Family expenses")

# 🔹 Pet Care
df["Pet_Care"] = safe_col(df, "Pet-care")

print("✅ Columns created successfully!")
```

```python
# 🔹 Fire / Property Insurance = Insurance + Rent + Utilities
df["Fire_Property_Insurance"] = df["Insurance"] + df["Rent"] + df["Utilities"]

# 🔹 Health Insurance = part of Insurance (assume 40%)
df["Health_Insurance"] = df["Insurance"] * 0.4  

# 🔹 Life Insurance = part of Insurance (assume 60%)
df["Life_Insurance"] = df["Insurance"] * 0.6  

# 🔹 Transport Related = Transport + Travel_Total
df["Transport_Expenses"] = df["Transport"] + df["Travel_Total"]

# 🔹 Accommodation = Rent + (Hotel if exists)
if "Hotel" in df.columns:
    df["Accommodation"] = df["Rent"] + df["Hotel"]
else:
    df["Accommodation"] = df["Rent"]

# 🔹 Family Related = Personal + Family_Total
if "Personal_Expenses" in df.columns:
    df["Family_Total"] = df["Personal_Expenses"] + df["Family_Total"]

# 🔹 Pet Care = Already exists, keep as is
df["Pet_Care_Final"] = df["Pet_Care"]
```

```python
df.head()
```

```python
# 🔹 Insurance Totals
df["Total_Insurance"] = (
    df["Fire_Property_Insurance"] +
    df["Health_Insurance"] +
    df["Life_Insurance"]
)

# 🔹 Travel Totals = Transport + Travel + Accommodation
df["Total_Travel"] = (
    df["Transport_Expenses"] +
    df["Accommodation"]
)

# 🔹 Family Totals = Family_Total (already) + Pet Care + Miscellaneous
df["Total_Family"] = (
    df["Family_Total"] +
    df["Pet_Care_Final"] +
    df["Miscellaneous"]
)

# 🔹 Grand Total (All grouped)
df["Grand_Total_Expenses"] = (
    df["Total_Insurance"] +
    df["Total_Travel"] +
    df["Total_Family"] +
    df["Groceries"] +
    df["Education"] +
    df["Entertainment"] +
    df["Eating_Out"]
)

# Check first rows
print(df[["Total_Insurance","Total_Travel","Total_Family","Grand_Total_Expenses"]].head())
```

```python
# ===============================
# 🔹 Create Chatbot-Ready Columns
# ===============================

# 1. Actual Savings
df["Actual_Savings"] = df["Income"] - (
    df["Rent"] + df["Loan_Repayment"] + df["Insurance"] + df["Groceries"] +
    df["Transport"] + df["Eating_Out"] + df["Entertainment"] +
    df["Utilities"] + df["Healthcare"] + df["Education"] +
    df["Miscellaneous"] + df["Travel_Total"] + df["Family_Total"] + df["Pet_Care"]
)

# 2. Expense Buckets
df["Housing_Expenses"] = df["Rent"] + df["Utilities"] + df["Loan_Repayment"]
df["Food_Expenses"] = df["Groceries"] + df["Eating_Out"]
df["Transport_Expenses"] = df["Transport"] + df["Travel_Total"]
df["Lifestyle_Expenses"] = df["Entertainment"] + df["Education"] + df["Miscellaneous"] + df["Family_Total"] + df["Pet_Care"]

# 3. Financial Health Indicators
df["Total_Expenses"] = (
    df["Housing_Expenses"] + df["Food_Expenses"] +
    df["Transport_Expenses"] + df["Lifestyle_Expenses"] + df["Insurance"]
)
df["Savings_Rate"] = (df["Actual_Savings"] / df["Income"]) * 100
df["Debt_to_Income_Ratio"] = (df["Loan_Repayment"] / df["Income"]) * 100
df["Expense_to_Income_Ratio"] = (df["Total_Expenses"] / df["Income"]) * 100

# 4. Flags for Recommendations
df["High_Spending_Flag"] = (df["Expense_to_Income_Ratio"] > 70).astype(int)
df["Low_Savings_Flag"] = (df["Savings_Rate"] < df["Desired_Savings_Percentage"]).astype(int)
df["High_Debt_Flag"] = (df["Debt_to_Income_Ratio"] > 40).astype(int)

# ✅ Preview new columns
df.head()


```
